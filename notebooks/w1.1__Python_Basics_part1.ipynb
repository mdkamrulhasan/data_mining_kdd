## 1. Variables and Basic Data Types

**Code cell purpose:** Create variables representing a simple patient record and inspect their Python data types.

````python
# Variables can store different kinds of values.
patient_id = "P-10432"          # str
age = 47                        # int
body_temp_c = 38.6              # float
is_fever = body_temp_c > 37.5   # bool
allergy_notes = None            # NoneType: no value recorded

print(patient_id, type(patient_id))
print(age, type(age))
print(body_temp_c, type(body_temp_c))
print(is_fever, type(is_fever))
print(allergy_notes, type(allergy_notes))
````

## 4. Conditional Statements

**Code cell purpose:** Define a reusable function that converts a numerical score into a letter grade using an `if`/`elif`/`else` chain.

````python
def letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

for score in [95, 82, 71, 58]:
    print(f"Score {score} -> Grade {letter_grade(score)}")
````

## 5. `for` Loops

**Code cell purpose:** Process a list of scores one item at a time and display the corresponding letter grade.

````python
scores = [95, 82, 71, 58, 88, 64, 90]

for score in scores:
    print(f"{score} -> {letter_grade(score)}")
````

**Code cell purpose:** Use `enumerate()` to access both the position and value of each item while processing a sequence.

````python
for i, score in enumerate(scores, start=1):
    print(f"Student #{i}: score {score}")
````

## 6. `while` Loops

**Code cell purpose:** Simulate sensor monitoring until three consecutive readings fall within a safe range or a maximum number of attempts is reached.

````python
# Predetermined readings make the example reproducible in Colab.
sensor_values = [45.2, 56.0, 92.4, 46.6, 50.8, 58.7]

consecutive_stable = 0
attempts = 0
readings = []

while consecutive_stable < 3 and attempts < len(sensor_values):
    value = sensor_values[attempts]
    readings.append(value)
    attempts += 1

    if 30.0 <= value <= 70.0:
        consecutive_stable += 1
    else:
        consecutive_stable = 0

print("Readings taken:", readings)
print("Attempts used:", attempts)
print("Stopped after 3 consecutive stable readings:", consecutive_stable == 3)
````

## 7. List Comprehensions

**Code cell purpose:** Use list comprehensions to transform scores, select passing scores, and apply a simple score adjustment.

````python
grades = [letter_grade(s) for s in scores]
passing_scores = [s for s in scores if s >= 60]
curved_passing = [s + 5 for s in scores if s >= 60]

print("Grades:", grades)
print("Passing scores:", passing_scores)
print("Curved passing scores:", curved_passing)
````

A common pattern is:

```python
[expression for item in sequence if condition]
```

This pattern is useful in KDD for selection, preprocessing, and transformation.

## 8. Functions

**Code cell purpose:** Define and call a simple function with one parameter and a return value.

````python
def celsius_to_fahrenheit(temp_c):
    """Convert a Celsius temperature to Fahrenheit."""
    return temp_c * 9 / 5 + 32

for temp_c in [0, 20, 37, 100]:
    print(f"{temp_c} C -> {celsius_to_fahrenheit(temp_c)} F")
````

## 9. Parameters, Default Arguments, and Return Values

**Code cell purpose:** Define a function with multiple parameters and a default argument to calculate a fixed loan payment.

````python
def loan_monthly_payment(principal, annual_rate_pct, years, periods_per_year=12):
    """Compute a fixed payment for a standard amortizing loan."""
    n_periods = years * periods_per_year
    rate_per_period = (annual_rate_pct / 100) / periods_per_year

    if rate_per_period == 0:
        return principal / n_periods

    payment = principal * (
        rate_per_period * (1 + rate_per_period) ** n_periods
    ) / ((1 + rate_per_period) ** n_periods - 1)

    return payment

print("Monthly payment:", round(loan_monthly_payment(20000, 6.0, 5), 2))
print("Quarterly payment:", round(
    loan_monthly_payment(20000, 6.0, 5, periods_per_year=4), 2
))
````

## 10. Returning Multiple Values

**Code cell purpose:** Calculate the minimum, maximum, and mean of a list using only basic Python and return all three values from one function.

````python
def summarize(values):
    """Return the minimum, maximum, and mean of a list of numbers."""
    total = 0
    lowest = values[0]
    highest = values[0]

    for v in values:
        total += v
        if v < lowest:
            lowest = v
        if v > highest:
            highest = v

    mean = total / len(values)
    return lowest, highest, mean

class_sizes = [22, 31, 18, 27, 25, 40, 19]
low, high, mean = summarize(class_sizes)

print("Minimum:", low)
print("Maximum:", high)
print("Mean:", round(mean, 2))
````

## 11. Function Composition

**Code cell purpose:** Reuse the previously defined `summarize()` function inside another function to identify oversized classes.

````python
def flag_oversized_classes(class_sizes, cap=30):
    _, largest, _ = summarize(class_sizes)
    oversized = [size for size in class_sizes if size > cap]
    over_by = largest - cap if largest > cap else 0
    return oversized, over_by

oversized, over_by = flag_oversized_classes(class_sizes, cap=30)
print("Oversized classes:", oversized)
print("Largest class exceeds the cap by:", over_by)
````

## 15. Sets

**Code cell purpose:** Compare two sets of topics using intersection, difference, and symmetric difference.

````python
section_a_topics = {"loops", "functions", "lists", "dictionaries"}
section_b_topics = {"functions", "dictionaries", "sets", "recursion"}

print("Both sections:", section_a_topics & section_b_topics)
print("A only:", section_a_topics - section_b_topics)
print("Exactly one section:", section_a_topics ^ section_b_topics)
````
